```python
total_sales = df['Gross Sales'].sum()
total_profit = df['Profit'].sum()
total_units = df['Units Sold'].sum()

print("Total Gross Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Units Sold:", total_units)

```

```python
country_sales = df.groupby('Country')['Gross Sales'].sum().sort_values(ascending=False)

print(country_sales)
```

```python
import matplotlib.pyplot as plt

country_sales.plot(kind='bar')

plt.title('Country-wise Gross Sales')
plt.xlabel('Country')
plt.ylabel('Gross Sales')
plt.xticks(rotation=45)

plt.show()
```

```python
product_sales = df.groupby('Product')['Gross Sales'].sum().sort_values(ascending=False)

print(product_sales)
```

```python
product_sales.plot(kind='bar')

plt.title('Product-wise Gross Sales')
plt.xlabel('Product')
plt.ylabel('Gross Sales')
plt.xticks(rotation=45)

plt.show()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month_name()

month_sales = df.groupby('Month')['Gross Sales'].sum()

print(month_sales)
```

```python
month_order = ['January', 'February', 'March', 'April', 'May', 'June',
               'July', 'August', 'September', 'October', 'November', 'December']

month_sales = month_sales.reindex(month_order)

plt.figure(figsize=(10,5))
plt.plot(month_sales.index, month_sales.values, marker='o')

plt.title('Month-wise Gross Sales Trend')
plt.xlabel('Month')
plt.ylabel('Gross Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
country_profit = df.groupby('Country')['Profit'].sum().sort_values(ascending=False)

print(country_profit)
```

```python
plt.figure(figsize=(8,5))
country_profit.plot(kind='bar')

plt.title('Country-wise Profit Analysis')
plt.xlabel('Country')
plt.ylabel('Profit')
plt.xticks(rotation=45)

plt.show()
```

```python
product_profit = df.groupby('Product')['Profit'].sum().sort_values(ascending=False)

print(product_profit)
```

```python
plt.figure(figsize=(8,5))
product_profit.plot(kind='bar')

plt.title('Product-wise Profit Analysis')
plt.xlabel('Product')
plt.ylabel('Profit')
plt.xticks(rotation=45)

plt.show()
```

```python
segment_analysis = df.groupby('Segment')[['Gross Sales', 'Profit']].sum()

print(segment_analysis)
```

```python
plt.figure(figsize=(8,5))

segment_analysis.plot(kind='bar')

plt.title('Segment-wise Gross Sales and Profit')
plt.xlabel('Segment')
plt.ylabel('Amount')
plt.xticks(rotation=0)
plt.legend(title='Metrics')

plt.show()
```

```python
discount_analysis = df.groupby('Discounts')[['Gross Sales', 'Profit']].mean().sort_index()

print(discount_analysis)
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    discount_analysis.index,
    discount_analysis['Profit'],
    marker='o'
)

plt.title('Impact of Discounts on Average Profit')
plt.xlabel('Discount')
plt.ylabel('Average Profit')
plt.xticks(rotation=45)

plt.show()
```
